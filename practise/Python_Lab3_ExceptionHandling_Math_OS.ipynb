# Exceptions: try/except, else, finally; Common Errors; raise

**Exercise 1:** Wrap a division `a/b` in try/except to catch `ZeroDivisionError` and print a helpful message.

```python
a,b = 8,0
try:
  output = a/b
  print(output)
except Exception as e:
  print(e)
```

**Exercise 2:** Convert the string `'42'` to int inside try/except/else/finally; print success in else, 'done' in finally.

```python
num = '42'
try:
  num = int(num)
except Exception as e:
  print(e)
else:
  print(f"Operation successfull\nresult: {type(num)}")
finally:
  print("Done")
```

**Exercise 3:** Open a missing file path and catch `FileNotFoundError` gracefully; print a friendly message.

```python
file_path = 'test.txt'
try:
  with open(file_path,"r") as doc:
    content = doc.read()
    print(content)
except FileNotFoundError:
  print("File not found on given location.\nKindly check your file location.")
```

**Exercise 8:** Given ISO date strings `['2025-05-01','2024-12-31','2025-01-15']`, sort them chronologically using `datetime`.

```python
dates_strings = ['2025-05-01','2024-12-31','2025-01-15']

dates_object = [datetime.strptime(d,"%Y-%m-%d") for d in dates_strings]

dates_object.sort()

sorted_dates = [d.strftime("%Y-%m-%d") for d in dates_object]

sorted_dates
```
