# 3. Palindrome-like Variation: Check reversed arring match ignoring case & non-alphabets.

```python
arr = 'maLA#$yalam'
brr = 'delhi'
```

```python
def check_palindrome(text):
  cleaned = "".join(ch.lower() for ch in text if ch.isalpha())

  return cleaned == cleaned[::-1]

check_palindrome(arr)
```

```python
check_palindrome(brr)
```

# 4. Character Frequency Finder: Build dict of char counts manually.

```python
def freq_count(text):
  freq = {}

  for ch in text:
    freq[ch] = freq.get(ch,0) + 1

  return freq
```

```python
freq_count('Honululu')
```

# 10. Unique Pairs: Create unique (a,b) pairs using list comprehension.

```python
nums = [1,2,3,4]
unique = []
length = len(nums)
for first in range(length-1):
  second = first + 1
  while second <  length:
    if nums[first] != nums[second]:
            unique.append((nums[first],nums[second]))
    second+=1
unique
```

# 14. Remove Duplicates with Sets: Maintain order while deduplicating.

```python
nums = [1,2,2,3,1,4,5,6,7,8,8,7,5,2]
seen = set()
result = []

for val in nums:
  if val not in seen:
    result.append(val)
    seen.add(val)

print(result)
```

# 15. Set Operations: Find only live, only recording, both, none.

```python
A = {"Alice", "Bob", "Charlie", "David"} # live
B = {"Charlie", "David", "Eve", "Frank"} # recording
print("Only Live:",A.difference(B))
print("Only Recording:", B.difference(A))
print("Both:",A.intersection(B))
print("None: ",A.union(B))
```

# 19. Nested Dictionary Access: Access dept, names, oldest age.

```python
company = {
    "HR": {
        "manager": "Alice",
        "employees": [
            {"name": "Noah", "age": 28},
            {"name": "Liam", "age": 32}
        ]
    },
    "IT": {
        "manager": "Bob",
        "employees": [
            {"name": "Oliver", "age": 25},
            {"name": "Mateo", "age": 45},
            {"name": "Elijah", "age": 30}
        ]
    }
}
```

```python
it_dept = company['IT']
hr_dept = company['HR']
```

```python
names = [emp['name']
         for dept in company.values() # get access of values in key(IT,HR).
         for emp in dept['employees'] # get access of values in employees.
        ]
names
```

```python
all_ages = [emp['age']
            for dept in company.values()
            for emp in dept['employees']]
print(all_ages)
max(all_ages)
```
